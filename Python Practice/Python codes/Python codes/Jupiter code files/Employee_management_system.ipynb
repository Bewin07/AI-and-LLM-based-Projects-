```python
class Employee_management_system:
    def __init__(self,username,password):
        self.username= username
        self.password=password

    def add_employee(self):
        print(f'choice selected: Adding employees.., please enter username and password to proceed.')
        if (self.username=="admin" and self.password==1212):
            n=int(input("enter the no.of employees to add: "))
            with open("employees.txt","a") as file:
                for i in range(n):
                    name=input('enter the name of the employee:')
                    file.write("\n"+name)
        else:
            print('Incorrect password')
    def view_employees(self):
        print(f'choice selected: Viewing all employees..')
        if (self.username=="admin" and self.password==1212):
            with open("employees.txt","r") as file:
                print(file.read())
        else:
            print('Incorrect password')
    def overwrite_employee(self):
        print(f'choice selected: Clearing all employees and adding new set of employees..')
        print(f'Are you sure want to clear all employee names and add new ones?')
        Yes_or_No= input('enter Yes or No:')
        if Yes_or_No.lower()=='yes':
            if (self.username=="admin" and self.password==1212):
                n=int(input("enter the no.of employees to add: "))
                with open("employees.txt","w") as file:
                    for i in range(n):
                        name=input('enter the name of the employee:')
                        file.write("\n"+name)
            else:
                print('Incorrect password')
        else:
            print('Ok, Thanks for confirmation')
    
```

```python
username=input("enter the username: ")
password=int(input('enter the password: '))
obj=Employee_management_system(username,password)
print(f'1. Add employees')
print(f'2. View employees')
print(f'3. Clear all employees and overwrite all employees')
choice=int(input('enter a choice 1 to 3: '))

if choice==1:
    obj.add_employee()
elif choice ==2:
    obj.view_employees()
elif choice==3:
    obj.overwrite_employee()
else:
    print("enter correct option between 1 to 3")
    
```
