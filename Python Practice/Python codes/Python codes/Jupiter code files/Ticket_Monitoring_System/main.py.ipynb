```python
import pandas as pd
import random
```

```python
types=['Incident','CR','Infra']
status=['in-progress','completed','yet-to-start']
priority=['High','Medium','Low']
owners=['Bewin','John','Mary','Ragu']
```

```python
data=[]
for i in  range(1,101):
    data.append({
        'Ticket ID':f'TKT{i:03}',
        'Ticket Type':random.choice(types),
        'Priority':random.choice(priority),
        'Assignee':random.choice(owners)
        
    })
df=pd.DataFrame(data)
df.to_excel('Tickets.xlsx',index=False)
```
